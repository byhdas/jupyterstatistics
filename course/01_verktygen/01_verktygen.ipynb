## 1.1 Python som miniräknare

Kör cellen nedan. Jupyter visar värdet av cellens **sista rad**.

````python
(172 + 165 + 180) / 3
````

Ett värde kan sparas i en **variabel**. Med `print` och en **f-sträng** (`f"..."`) skriver du ut text med värden i.
Efter kolonet anger du hur talet ska visas: `.1f` betyder en decimal och `.0%` betyder procent utan decimaler.

````python
mean_height = (172 + 165 + 180) / 3
share = 2 / 3

print(f"Medellängd: {mean_height:.1f} cm")
print(f"Andel: {share:.0%}")
````

## 1.2 NumPy: räkna med många tal på en gång

I statistik räknar vi nästan aldrig på ett tal i taget. En NumPy-**array** är en talserie där en beräkning
görs på alla värden samtidigt.

````python
import numpy as np

heights = np.array([172, 165, 180, 158, 191])

print(heights - 170)        # dra 170 från alla värden
print(heights.mean())       # medelvärde
print(heights.max() - heights.min())
````

### Ett trick som används i hela kursen: villkor blir andelar

Ett villkor på en array ger `True` eller `False` för varje värde. Python räknar `True` som 1 och `False` som 0.
Därför ger **summan** antalet som uppfyller villkoret, och **medelvärdet** ger andelen.

🔮 **Gissa först:** vad skriver de tre raderna ut?

````python
print(heights > 170)
print((heights > 170).sum())
print((heights > 170).mean())
````

## 1.3 Slumptal och frön

All data i kursen är **simulerad**: vi låter datorn slumpa fram den. Fördelen är att vi då känner till
"sanningen" och kan se hur bra statistiken lyckas hitta den.

En slumptalsgenerator skapas med ett **frö** (seed). Samma frö ger samma "slump" varje gång,
så att dina resultat går att upprepa.

````python
rng = np.random.default_rng(42)     # ← ändra fröet 42 till något annat

dice = rng.integers(1, 7, 10)       # 10 tärningskast: heltal från 1 till 6
print(dice)
print(f"Medelvärde: {dice.mean():.2f}")
````

Fler slumpfunktioner som dyker upp längre fram:

| Kod | Ger |
|---|---|
| `rng.normal(172, 9, 5)` | 5 tal kring 172 med standardavvikelse 9 (t.ex. längder) |
| `rng.uniform(0, 20, 5)` | 5 tal jämnt utspridda mellan 0 och 20 |
| `rng.choice(["A", "B"], 5, p=[0.7, 0.3])` | 5 slumpvisa val, A med 70 % sannolikhet |

## 1.4 pandas: data i tabeller

En **DataFrame** är en tabell, ungefär som ett kalkylblad. Varje rad är en individ och varje kolumn en variabel.

````python
import pandas as pd

people = pd.DataFrame({
    "name": ["Ada", "Bo", "Cia", "Dan", "Eva", "Fia"],
    "height": [165, 180, 172, 191, 158, 169],
    "city": ["Lund", "Umeå", "Lund", "Lund", "Umeå", "Visby"],
})
people
````

Tre saker du gör hela tiden: välja en kolumn, räkna på den och filtrera rader.

````python
print(people["height"].mean())          # en kolumn och dess medelvärde
print(people["city"].value_counts())    # hur många gånger varje värde förekommer
````

````python
people[people["height"] > 170]          # bara raderna där villkoret är sant
````

## 1.5 matplotlib: diagram

Ett diagram säger ofta mer än en tabell. Ett **histogram** visar hur värdena fördelar sig.
Sätt alltid rubriker på axlarna.

````python
import matplotlib.pyplot as plt

sim_data = rng.normal(172, 9, 500)       # 500 simulerade längder

plt.hist(sim_data, bins=20, edgecolor="white")       # ← ändra bins
plt.axvline(sim_data.mean(), color="red", linestyle="--", label="medelvärde")
plt.xlabel("Längd (cm)")
plt.ylabel("Antal")
plt.title("500 simulerade längder")
plt.legend()
plt.show()
````

## Sammanfattning

- En notebook är till för att **experimentera**: ändra, kör om, jämför.
- NumPy räknar på hela talserier. `(villkor).mean()` ger en **andel**.
- `rng = np.random.default_rng(42)` ger slump som går att upprepa.
- pandas håller data i tabeller: `df["kolumn"]`, `df[villkor]`, `.mean()`, `.value_counts()`.
- matplotlib ritar: `plt.hist(...)`, och glöm inte `plt.xlabel` och `plt.ylabel`.
